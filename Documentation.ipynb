## Register File
Finished implementing the register desiged around SIMT execution

### input params
The register has three inputs
- THREADS - representing the number of threads
- REGS - representing the number of regfiles we want
- DATA_W - the width of our data in bits

### Wires
The register has the following virtual wiring
- clk - to control the clock signal
- we - to enable writes
- rd - a desitnation register
- rs1 - a source register 
- rs2 - a second source register
- rs1_data - a bus for data streamed from rs1
- rs2_data - a bus for data streamed from rs2
- rd_data - a bus for data streamed from rd

### Internal Logic
We define the regfile space as THREADS * REGS registers each DATA_W bits wide

The notation is regfile[thread][reg #]

We then use `generate` to enable compiler-time replication - creating THREAD copies of combinational logic to assign the databus inputs to the correct registers in the reg file

We also have writeback logic in this case sequential on the clock edge. When we is true we employ an activemasking to only write to desired registers
